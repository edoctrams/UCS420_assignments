```python
import numpy as np


arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("a) Reversed array:", reversed_arr)



def find_most_frequent(arr_input):

    counts = np.bincount(arr_input)
    most_frequent_val = np.argmax(counts)
    indices = np.where(arr_input == most_frequent_val)[0]
    return most_frequent_val, indices



x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
val_x, idx_x = find_most_frequent(x)
print(f"b) i. For array x: Most frequent value is {val_x} at indices {idx_x}")


y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

counts_y = np.bincount(y)
max_count = np.max(counts_y)
modes = np.where(counts_y == max_count)[0]

for mode in modes:
    idx_y = np.where(y == mode)[0]
    print(
        f"b) ii. For array y: Most frequent value is {mode} at indices {idx_y}"
    )
```

```python
import numpy as np


your_name = np.linspace(10, 100, 25)
print("Array details:")
print("Dimensions (ndim):", your_name.ndim)
print("Shape:", your_name.shape)
print("Total elements (size):", your_name.size)
print("Data type (dtype):", your_name.dtype)
print("Total bytes consumed:", your_name.nbytes)



reshaped_transpose = your_name.reshape(-1, 1)
print("\nTranspose using reshape(-1, 1):\n", reshaped_transpose)


print("\nCan we use .T on 1D directly?")
print(
    "No, using your_name.T on a true 1D array returns the exact same 1D array: ",
    your_name.T.shape,
)
print(
    "To make .T work like a matrix transpose, we must first add an axis, e.g., your_name[np.newaxis, :].T"
)
```
