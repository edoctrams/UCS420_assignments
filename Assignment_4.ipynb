```python
import pandas as pd


fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"},
]
custom_entries = [
    {"question": "where is your office located", "answer": "Our main office is located in Tech Park, Delhi.", "keywords": "location office address place", "category": "general"},
    {"question": "how do i close my account", "answer": "Go to Account Settings and click Close Account.", "keywords": "close delete terminate account", "category": "account"}
]
df = pd.DataFrame(fixed_entries + custom_entries)



def simple_search(query, dataframe):

    query_words = query.lower().split()


    scores = []

    for index, row in dataframe.iterrows():
        match_count = 0


        for word in query_words:
            if word in row['keywords']:
                match_count += 1

        scores.append(match_count)


    results_df = dataframe.copy()
    results_df['score'] = scores

    matched_results = results_df[results_df['score'] > 0]
    return matched_results.sort_values(by='score', ascending=False)



user_query = "how to pay my fee"
output = simple_search(user_query, df)

print(output[['score', 'question', 'answer']])
```

```python
import pandas as pd

df = pd.DataFrame([
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"},
    {"question": "where is your office located", "answer": "Our main office is located in Tech Park, Delhi.", "keywords": "location office address place", "category": "general"},
    {"question": "how do i close my account", "answer": "Go to Account Settings and click Close Account.", "keywords": "close delete terminate account", "category": "account"}
])

def search_with_ties(query, dataframe):
    query_words = query.lower().split()
    scores = []

    for index, row in dataframe.iterrows():
        match_count = 0
        for word in query_words:
            if word in row['keywords']:
                match_count += 1
        scores.append(match_count)

    results_df = dataframe.copy()
    results_df['score'] = scores


    matched_results = results_df[results_df['score'] > 0]

    if matched_results.empty:
        print("No matches found.")
        return matched_results


    highest_score = matched_results['score'].max()


    top_matches = matched_results[matched_results['score'] == highest_score]

    print(f"\nHighest Match Score: {highest_score}")
    print(top_matches[['score', 'question', 'answer']])
    return top_matches



print("==================================================")
print("TEST 1: Querying 'fee' (Should return a tie of 2 rows)")
print("==================================================")
search_with_ties("fee", df)


print("\n==================================================")
print("TEST 2: Querying 'reset login' (Should return 1 clear winner)")
print("==================================================")
search_with_ties("reset login", df)
```
